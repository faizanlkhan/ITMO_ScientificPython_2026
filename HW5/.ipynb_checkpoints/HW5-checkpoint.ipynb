# Heart Disease Dataset — Exploratory Data Analysis

Heart disease is still the leading cause of death globally, and being able to predict it early can genuinely save lives. This dataset was put together from five different medical datasets (Cleveland, Hungarian, Switzerland, Long Beach VA, and Stalog) and is available on Kaggle. The combined version has 918 patient records, which is a decent sample to work with.

The goal here is to understand what's inside the data before any modeling — who's in it, what the features mean, where the data has problems, and which variables seem to matter most for predicting heart disease.

Let's dig in.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 110
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
```

```python
df = pd.read_csv('heart.csv')

print(f'Shape: {df.shape}')
print()
print(df.dtypes)
```

```python
df.head(10)
```

## Understanding the Features

Before jumping into plots, it's worth taking a moment to understand what each column actually represents. Some of these are straightforward, others need a bit of medical context.

| Feature | What it is |
|---|---|
| **Age** | Patient age in years |
| **Sex** | M = Male, F = Female |
| **ChestPainType** | Type of chest pain: ATA (Atypical Angina), NAP (Non-Anginal Pain), ASY (Asymptomatic), TA (Typical Angina) |
| **RestingBP** | Resting blood pressure in mm Hg |
| **Cholesterol** | Serum cholesterol in mg/dL |
| **FastingBS** | Fasting blood sugar — 1 if > 120 mg/dL, 0 otherwise |
| **RestingECG** | Resting electrocardiogram results: Normal, ST (ST-T wave abnormality), LVH (left ventricular hypertrophy) |
| **MaxHR** | Maximum heart rate achieved during exercise |
| **ExerciseAngina** | Whether exercise caused chest pain (Y/N) |
| **Oldpeak** | ST depression induced by exercise relative to rest |
| **ST_Slope** | Slope of the peak exercise ST segment: Up, Flat, Down |
| **HeartDisease** | Target — 1 means heart disease, 0 means normal |

A quick note on a few of these: ST depression (Oldpeak) and the ST slope are measurements from an ECG — they reflect how the heart's electrical activity changes during stress. Asymptomatic chest pain (ASY) sounds counterintuitive but it's actually a red flag; it means the patient has signs consistent with angina but doesn't feel the typical pain, which can indicate a more severe condition.

## Missing Data and Data Quality

First thing to check — are there any actual nulls?

```python
df.isnull().sum()
```

No missing values. But that doesn't mean the data is clean — it just means there are no explicit NaN cells. Let's look a bit closer at the numeric columns.

```python
df.describe().round(2)
```

A few things stand out immediately:

- **RestingBP minimum is 0** — a blood pressure of zero is not possible for a living person. This is almost certainly an entry error.
- **Cholesterol minimum is also 0** — again, biologically not possible. A cholesterol reading of 0 means the measurement wasn't recorded and was filled with zero instead of being left as null.
- **Oldpeak goes negative (min = -2.6)** — technically Oldpeak can be slightly negative depending on how the measurement is taken, but it's worth flagging.

```python
print('Rows with Cholesterol == 0:', (df['Cholesterol'] == 0).sum())
print('Rows with RestingBP == 0:', (df['RestingBP'] == 0).sum())
print('Rows with Oldpeak < 0:', (df['Oldpeak'] < 0).sum())
```

So 172 out of 918 rows have Cholesterol = 0 — that's nearly 19% of the dataset. These are effectively missing values. For now we'll keep them in and flag them, but any model trained on this data should handle them carefully (median imputation by age group or sex would be a reasonable approach).

The single RestingBP = 0 is likely a typo. And the 13 negative Oldpeak values are technically within the range that some ECG interpretations allow, so we'll leave those.

```python
df['chol_missing'] = (df['Cholesterol'] == 0).astype(int)

print('Heart disease rate when cholesterol IS recorded:')
print(df[df['chol_missing'] == 0]['HeartDisease'].mean().round(3))
print()
print('Heart disease rate when cholesterol is missing (= 0):')
print(df[df['chol_missing'] == 1]['HeartDisease'].mean().round(3))
```

Interesting — the missing cholesterol group has a higher heart disease rate. This suggests the zeros aren't random; patients with heart disease may have had cholesterol measurements skipped or recorded differently. This is a pattern worth keeping in mind.

## Target Variable

How balanced is our dataset?

```python
counts = df['HeartDisease'].value_counts()
labels = ['No Heart Disease', 'Heart Disease']

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(labels, [counts[0], counts[1]], color=['#5B8DB8', '#C0504D'], width=0.5, edgecolor='white')
axes[0].set_ylabel('Count')
axes[0].set_title('Target Distribution')
for i, v in enumerate([counts[0], counts[1]]):
    axes[0].text(i, v + 5, str(v), ha='center', fontsize=11)

axes[1].pie([counts[0], counts[1]], labels=labels, colors=['#5B8DB8', '#C0504D'],
            autopct='%1.1f%%', startangle=90, wedgeprops={'edgecolor': 'white', 'linewidth': 1.5})
axes[1].set_title('Target Split')

plt.tight_layout()
plt.show()
```

410 patients without heart disease (44.7%) vs 508 with it (55.3%). Slightly more positives than negatives, but it's not a severe imbalance — most classifiers should handle this without special treatment, though it's something to keep in mind when evaluating model performance.

## Univariate Analysis — Numeric Features

Let's look at the distributions of the continuous variables before bringing in the target.

```python
num_cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    data = df[col] if col != 'Cholesterol' else df[df['Cholesterol'] > 0]['Cholesterol']
    axes[i].hist(data, bins=30, color='#5B8DB8', edgecolor='white', linewidth=0.6)
    axes[i].set_title(col)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Count')

axes[5].set_visible(False)

fig.suptitle('Distribution of Numeric Features (Cholesterol excludes zeros)', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

A few observations:

- **Age** is fairly normally distributed, centered around 54. The bulk of patients are between 45 and 65.
- **RestingBP** looks roughly normal but has a long right tail — there are some patients with quite high blood pressure (180-200 range).
- **Cholesterol** (excluding the zero entries) is right-skewed with a couple of extreme values above 500. Anything above ~400 mg/dL would be considered very high clinically.
- **MaxHR** is left-skewed — most patients achieve a max heart rate of 120-170, with fewer at the extremes.
- **Oldpeak** is heavily right-skewed with most values near zero, but some patients showing significant ST depression.

## Univariate Analysis — Categorical Features

```python
cat_cols = ['Sex', 'ChestPainType', 'FastingBS', 'RestingECG', 'ExerciseAngina', 'ST_Slope']
cat_labels = {
    'FastingBS': {0: 'Normal', 1: '> 120 mg/dL'}
}

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()

colors = ['#5B8DB8', '#C0504D', '#70AD47', '#ED7D31']

for i, col in enumerate(cat_cols):
    vc = df[col].value_counts()
    axes[i].bar(vc.index.astype(str), vc.values, color=colors[:len(vc)], edgecolor='white')
    axes[i].set_title(col)
    axes[i].set_ylabel('Count')
    for j, v in enumerate(vc.values):
        axes[i].text(j, v + 3, str(v), ha='center', fontsize=9)

fig.suptitle('Categorical Feature Counts', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

The dataset is heavily male — 725 men vs 193 women. This is common in cardiovascular studies (historically most clinical trials enrolled more men), but it's a limitation. Any gender-related findings need to be interpreted carefully given the imbalance.

Asymptomatic chest pain (ASY) is the most frequent type by far. Most patients have a normal resting ECG. And only about 23% of patients have elevated fasting blood sugar.

## Feature vs Heart Disease — Numeric Variables

Now the interesting part — how do these features differ between patients who have heart disease and those who don't?

```python
df['HeartDisease_label'] = df['HeartDisease'].map({0: 'No Disease', 1: 'Heart Disease'})

fig, axes = plt.subplots(2, 3, figsize=(14, 9))
axes = axes.flatten()

palette = {'No Disease': '#5B8DB8', 'Heart Disease': '#C0504D'}

for i, col in enumerate(num_cols):
    plot_df = df.copy()
    if col == 'Cholesterol':
        plot_df = df[df['Cholesterol'] > 0].copy()
    sns.boxplot(data=plot_df, x='HeartDisease_label', y=col,
                palette=palette, ax=axes[i], width=0.5, linewidth=1.2,
                flierprops={'marker': 'o', 'markersize': 3, 'alpha': 0.5})
    axes[i].set_title(col)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('')

axes[5].set_visible(False)
fig.suptitle('Numeric Features by Heart Disease Status', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

This is where things get clear:

- **Age**: Heart disease patients are noticeably older on average (~56 vs ~51). The spread is similar but the median shifts up meaningfully.
- **MaxHR**: One of the strongest signals — patients with heart disease have a much lower maximum heart rate during exercise. The medians are separated by roughly 20 bpm.
- **Oldpeak**: Higher in the heart disease group, which makes clinical sense — more ST depression indicates the heart is under more stress during exercise.
- **RestingBP**: Small difference, not a very clean separator on its own.
- **Cholesterol** (non-zero only): Interestingly, the healthy group actually shows higher cholesterol here. This is counterintuitive at first, but may partly be because patients with known heart disease are more likely to be on statins (cholesterol-lowering medication), which would lower their measured levels.

## Feature vs Heart Disease — Categorical Variables

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 9))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    ct = pd.crosstab(df[col], df['HeartDisease_label'], normalize='index') * 100
    ct.plot(kind='bar', ax=axes[i], color=['#5B8DB8', '#C0504D'],
            edgecolor='white', width=0.6, rot=0)
    axes[i].set_title(col)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('% within group')
    axes[i].legend(title='', fontsize=8)
    axes[i].yaxis.set_major_formatter(ticker.FormatStrFormatter('%.0f%%'))

fig.suptitle('Heart Disease Rate by Categorical Feature', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

Very telling patterns here:

- **Sex**: About 63% of male patients in this dataset have heart disease, compared to only 26% of female patients. Though remember the dataset has far more men, so this isn't saying women are protected — it's partly a reflection of the dataset composition.
- **ChestPainType**: ASY (asymptomatic) is the most dangerous — nearly 79% of ASY patients have heart disease. Atypical angina (ATA) is actually the safest group (~14% disease rate), which tells us classic-looking chest pain in this dataset is paradoxically less worrying than no pain at all.
- **ExerciseAngina**: If exercise causes chest pain, ~85% of those patients have heart disease. Very strong predictor.
- **ST_Slope**: Flat or downward-sloping ST segments are strongly associated with disease (~83% and ~78%). An upward slope is generally considered healthy — only ~20% of those patients have disease.
- **FastingBS**: Elevated fasting blood sugar shows a somewhat higher disease rate, consistent with diabetes being a cardiovascular risk factor.
- **RestingECG**: LVH and ST abnormalities both show higher disease rates than Normal, but the separation isn't as clean as with some other features.

## Correlation Analysis

```python
encoded = df.copy()
encoded['Sex_M'] = (df['Sex'] == 'M').astype(int)
encoded['ExerciseAngina_Y'] = (df['ExerciseAngina'] == 'Y').astype(int)
encoded['ChestPain_ASY'] = (df['ChestPainType'] == 'ASY').astype(int)
encoded['ST_Flat'] = (df['ST_Slope'] == 'Flat').astype(int)

corr_cols = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak',
             'FastingBS', 'Sex_M', 'ExerciseAngina_Y', 'ChestPain_ASY', 'ST_Flat', 'HeartDisease']

corr = encoded[corr_cols].corr()

fig, ax = plt.subplots(figsize=(11, 9))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, ax=ax,
            linewidths=0.5, annot_kws={'size': 8})
ax.set_title('Correlation Matrix (includes encoded categoricals)', fontsize=12)
plt.tight_layout()
plt.show()
```

Looking at the HeartDisease column specifically — ExerciseAngina, ST_Flat, ChestPain_ASY, and Sex_M all show positive correlations with the target, while MaxHR is negatively correlated (higher max heart rate → less likely to have disease). Oldpeak and Age also correlate positively.

There's also a notable negative correlation between Age and MaxHR — older patients naturally achieve lower maximum heart rates, which makes sense physiologically.

There's a visible pattern here — the red dots (heart disease) tend to cluster in the older ages with lower max heart rates, while the blue dots (no disease) are more spread toward higher max heart rates across all ages. The boundary isn't perfectly clean, but the separation is real.

The left plot confirms what the bivariate analysis suggested — Flat and Down slopes come with higher Oldpeak values, especially in heart disease patients. And looking at the right plot, for ASY chest pain the heart disease group tends to be older, while for other chest pain types the age distributions overlap more.

## Outlier Check

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 5))

for ax, col in zip(axes, ['RestingBP', 'Cholesterol', 'Oldpeak']):
    sns.boxplot(y=df[col], ax=ax, color='#5B8DB8', width=0.4,
                flierprops={'marker': 'o', 'markersize': 4, 'markerfacecolor': '#C0504D', 'alpha': 0.7})
    ax.set_title(col)

fig.suptitle('Outlier Inspection', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
for col in ['RestingBP', 'Cholesterol', 'Oldpeak']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    low = Q1 - 1.5 * IQR
    high = Q3 + 1.5 * IQR
    n = ((df[col] < low) | (df[col] > high)).sum()
    print(f'{col}: {n} potential outliers (IQR method) — [{low:.1f}, {high:.1f}]')
```

Cholesterol has the most outliers by IQR — some of those values above 500 mg/dL are quite extreme and could be genuine cases of familial hypercholesterolemia, or they could be measurement errors. Hard to tell without clinical context. Oldpeak's outliers on the high end are also notable but less surprising given how the feature is distributed.

## Feature Engineering Suggestions

Based on what we've seen, here are some thoughts on how to prepare this data for modeling:

**Encoding categorical variables:**
- `Sex`, `ExerciseAngina` are binary — simple 0/1 encoding works fine
- `ChestPainType`, `RestingECG`, `ST_Slope` have 3-4 categories each — one-hot encoding would be cleaner than label encoding since there's no natural ordinal relationship (Flat ST slope isn't "more" than Up, it's just different)

**Handling Cholesterol zeros:**
- The 172 zero values shouldn't be fed to a model as-is. Options:
  - Impute with median cholesterol (either overall or grouped by sex/age group)
  - Add a binary flag column `chol_is_missing` to let the model know these were imputed
  - Or simply drop those rows if the dataset size allows it

**New features worth trying:**
- Age bins (e.g., < 45, 45–55, 55–65, 65+) might capture non-linear age effects better than raw age
- A combined ST feature using both Oldpeak magnitude and ST_Slope direction
- The RestingBP zero row should just be dropped or imputed

**Scaling:**
- Age, RestingBP, MaxHR, Oldpeak, Cholesterol are all on different scales — standardization or min-max scaling would be needed for distance-based models (KNN, SVM, neural nets). Tree-based models don't need it.

## Summary and Conclusions

After going through the data, here's what stands out:

**The dataset is usable but not perfectly clean.** The 172 missing cholesterol values (encoded as zeros) are the biggest quality issue and need to be addressed before modeling. Everything else is in reasonable shape.

**The strongest predictors of heart disease in this dataset appear to be:**
1. **ST_Slope** — Flat or Down slope is a very strong signal
2. **ExerciseAngina** — if exercise triggers chest pain, the disease rate jumps to ~85%
3. **ChestPainType** — counterintuitively, asymptomatic patients have the highest disease rate
4. **MaxHR** — lower max heart rate during exercise is consistently associated with disease
5. **Oldpeak** — more ST depression means more cardiac stress
6. **Age** and **Sex** are also relevant but less powerful on their own

**What's not as useful as you might expect:** Resting blood pressure on its own is a fairly weak separator. Cholesterol is similarly not a clean predictor — and the zero-value issue makes it harder to use.

The dataset looks well-suited for a classification task. Given the feature distributions and the relatively balanced target, a gradient boosting model or random forest would probably be a good starting point — they handle mixed feature types well and don't require the categorical variables to be encoded in any particular way.
